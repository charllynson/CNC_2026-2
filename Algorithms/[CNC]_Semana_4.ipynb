### **Algoritmo 1 - Algoritmo Eliminação de Gauss sem pivotamento - Ruggiero**

```python
A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b = [8, -11, -3]

n = len(A)

for k in range(n - 1):

    for i in range(k + 1, n):

        m = A[i][k] / A[k][k]

        A[i][k] = 0

        for j in range(k + 1, n):
            A[i][j] = A[i][j] - m * A[k][j]

        b[i] = b[i] - m * b[k]


x = [0] * n

x[n - 1] = b[n - 1] / A[n - 1][n - 1]

for k in range(n - 2, -1, -1):

    s = 0

    for j in range(k + 1, n):
        s = s + A[k][j] * x[j]

    x[k] = (b[k] - s) / A[k][k]


print("Solução:")
for i in range(n):
    print("x", i + 1, "=", x[i])
```

### **Algoritmo 2 - Algoritmo Eliminação de Gauss - Filho**

```python
A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b = [8, -11, -3]

n = len(A)

det = 1
info = 0

for j in range(n - 1):

    p = j
    Amax = abs(A[j][j])

    for k in range(j + 1, n):
        if abs(A[k][j]) > Amax:
            Amax = abs(A[k][j])
            p = k

    if p != j:
        for k in range(n):
            t = A[j][k]
            A[j][k] = A[p][k]
            A[p][k] = t

        t = b[j]
        b[j] = b[p]
        b[p] = t

        det = -det

    det = det * A[j][j]

    if abs(A[j][j]) != 0:

        r = 1 / A[j][j]

        for i in range(j + 1, n):

            Mult = A[i][j] * r
            A[i][j] = 0

            for k in range(j + 1, n):
                A[i][k] = A[i][k] - Mult * A[j][k]

            b[i] = b[i] - Mult * b[j]

    else:
        if info == 0:
            info = j + 1

det = det * A[n - 1][n - 1]

if info == 0 and abs(A[n - 1][n - 1]) == 0:
    info = n


print("Matriz A:")
for linha in A:
    print(linha)

print("\nVetor b:")
print(b)

print("\nDeterminante:", det)
print("Info:", info)
```

### **Algoritmo 3 - Algoritmo Fatoração LU - Filho**

```python
A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

m = len(A)
n = len(A[0])

Pivot = []

for i in range(m):
    Pivot.append(i)

PdU = 1
info = 0

for j in range(min(m, n)):

    p = j
    Amax = abs(A[j][j])

    for k in range(j + 1, m):
        if abs(A[k][j]) > Amax:
            Amax = abs(A[k][j])
            p = k

    if p != j:

        for k in range(n):
            t = A[j][k]
            A[j][k] = A[p][k]
            A[p][k] = t

        i = Pivot[j]
        Pivot[j] = Pivot[p]
        Pivot[p] = i

        PdU = -PdU

    PdU = PdU * A[j][j]

    if abs(A[j][j]) != 0:

        r = 1 / A[j][j]

        for i in range(j + 1, m):

            Mult = A[i][j] * r
            A[i][j] = Mult

            for k in range(j + 1, n):
                A[i][k] = A[i][k] - Mult * A[j][k]

    else:

        if info == 0:
            info = j + 1


print("Matriz LU:")

for linha in A:
    print(linha)

print("\nPivôs:")
print(Pivot)

print("\nProduto da diagonal de U:", PdU)

print("\nInfo:", info)
```
