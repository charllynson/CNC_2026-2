### **Algoritmo 1- Algoritmo do Método da Bissecção**

```python
def f(x):
    return x**3 - x - 2

a = 1
b = 2
e = 0.0001

M = f(a)
k = 1

if (b - a) < e:
    x = (a + b) / 2

else:
    while (b - a) >= e:
        x = (a + b) / 2

        if M * f(x) > 0:
            a = x
        else:
            b = x

        if (b - a) < e:
            x = (a + b) / 2
            break

        k = k + 1

print("Raiz aproximada:", x)
print("Iterações:", k)
```

### **Algoritmo 2 - Algoritmo do Método da Falsa Posição**

```python
def f(x):
  return x**3 - x - 2

a = 1
b = 2
e1 = 0.0001
e2 = 0.0001

k = 1
M = f(a)

while True:
  x = (a * f(b) - b * f(a)) / (f(b) - f(a))

  if abs(f(x)) < e2:
    print("Raiz aproximada: ", x)
    break

  if M * f(x) > 0:
    a = x
    M = f(a)
  else:
    b = x

  if (b - a) < e1:
    print("Raiz aproximada: ", x)
    break
  k = k + 1

print("Iterações: ", k)
```
